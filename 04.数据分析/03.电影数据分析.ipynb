```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.core.pylabtools import figsize
from matplotlib.lines import lineStyles
from matplotlib.pyplot import figure
from numpy import dtype

#展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']
#创建子图
fig, ax = plt.subplots(figsize = (20, 12),dpi = 100)
fig.suptitle('电影数据分析', fontsize = 30)
#获取子图
ax1 : plt.Axes = ax
ax1.set_title('电影年份折线图', fontsize = 30)
#读取数据
df = pd.read_csv('data/movie.csv')
#获取年份
year_x = df.groupby('年份')['年份'].count()

year_min = year_x.index.min()

year_max = year_x.index.max()
#x轴
x = [i for i in range(year_min, year_max + 1)]
#y轴
y = [int(year_x.get(i, 0)) for i in x]
#绘制
ax1.plot(x, y)
ax1.set_xlabel('年份')
ax1.set_ylabel('电影数量')
ax1.set_xticks(x[::10])
y_ticks = [i for i in range(0, 10, 1)]
ax1.set_yticks(y_ticks)
ax1.grid(linestyle='--',alpha=0.3)
```
